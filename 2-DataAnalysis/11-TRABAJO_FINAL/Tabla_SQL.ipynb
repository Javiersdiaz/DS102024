```python
CREATE TABLE actividades (                                 
    id_evento SERIAL PRIMARY KEY,
    nombre_evento VARCHAR(255) NOT NULL,
    id_categoria INT NOT NULL,
    id_discapacidad INT,
    fecha DATE,
    id_ubicacion INT,
    costo DECIMAL(10, 2),
    edad_min INT,
    edad_max INT,
    min_integrantes INT,
    max_integrantes INT,
    id_modalidad INT,
    descripcion TEXT,
    activo BOOLEAN DEFAULT TRUE,
    origen_web TEXT,
    ultimo_scrape TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
    FOREIGN KEY (id_categoria) REFERENCES categorias(id_categoria),
    FOREIGN KEY (id_discapacidad) REFERENCES discapacidades(id_discapacidad),
    FOREIGN KEY (id_ubicacion) REFERENCES ubicaciones(id_ubicacion),
    FOREIGN KEY (id_modalidad) REFERENCES modalidades(id_modalidad)
);

```

```python
CREATE TABLE categorias (
    id_categoria SERIAL PRIMARY KEY,
    nombre_categoria VARCHAR(100) NOT NULL
);
```

```python
CREATE TABLE discapacidades (
    id_discapacidad SERIAL PRIMARY KEY,
    tipo VARCHAR(100) NOT NULL
);
```

```python
CREATE TABLE modalidades (
    id_modalidad SERIAL PRIMARY KEY,
    descripcion VARCHAR(100) NOT NULL
);
```

```python
CREATE TABLE ubicaciones (
    id_ubicacion SERIAL PRIMARY KEY,
    nombre_lugar VARCHAR(255) NOT NULL,
    direccion TEXT,
    codigo_postal VARCHAR(20),
    distrito VARCHAR(100),
    latitud DECIMAL(9, 6),
    longitud DECIMAL(9, 6)
);
```


1. Tabla actividades es la tabla principal.

2. Cada actividad se relaciona con una categoría en la tabla categorias.

3. Las actividades pueden tener un tipo de discapacidad (relacionado con discapacidades).

4. Se asocia con una ubicación específica (relacionado con ubicaciones).

5. Cada actividad tiene una modalidad (relacionado con modalidades).

```python
import psycopg2

# Conexión a la base de datos
conn = psycopg2.connect(
    dbname="actividades_ninos",  # Cambia por el nombre de tu base de datos
    user="postgres",             # Cambia por tu usuario
    password="tu_contraseña",   # Cambia por tu contraseña
    host="localhost",
    port="5432"
)
cur = conn.cursor()

# Crear tablas
queries = [
    """
    CREATE TABLE IF NOT EXISTS categorias (
        id_categoria SERIAL PRIMARY KEY,
        nombre_categoria VARCHAR(100) NOT NULL
    );
    """,
    """
    CREATE TABLE IF NOT EXISTS discapacidades (
        id_discapacidad SERIAL PRIMARY KEY,
        tipo VARCHAR(100) NOT NULL
    );
    """,
    """
    CREATE TABLE IF NOT EXISTS modalidades (
        id_modalidad SERIAL PRIMARY KEY,
        descripcion VARCHAR(100) NOT NULL
    );
    """,
    """
    CREATE TABLE IF NOT EXISTS ubicaciones (
        id_ubicacion SERIAL PRIMARY KEY,
        nombre_lugar VARCHAR(255) NOT NULL,
        direccion TEXT,
        codigo_postal VARCHAR(20),
        distrito VARCHAR(100),
        latitud DECIMAL(9, 6),
        longitud DECIMAL(9, 6)
    );
    """,
    """
    CREATE TABLE IF NOT EXISTS actividades (
        id_evento SERIAL PRIMARY KEY,
        nombre_evento VARCHAR(255) NOT NULL,
        id_categoria INT NOT NULL,
        id_discapacidad INT,
        fecha DATE,
        id_ubicacion INT,
        costo DECIMAL(10, 2),
        edad_min INT,
        edad_max INT,
        min_integrantes INT,
        max_integrantes INT,
        id_modalidad INT,
        descripcion TEXT,
        activo BOOLEAN DEFAULT TRUE,
        origen_web TEXT,
        ultimo_scrape TIMESTAMP DEFAULT CURRENT_TIMESTAMP,
        FOREIGN KEY (id_categoria) REFERENCES categorias(id_categoria),
        FOREIGN KEY (id_discapacidad) REFERENCES discapacidades(id_discapacidad),
        FOREIGN KEY (id_ubicacion) REFERENCES ubicaciones(id_ubicacion),
        FOREIGN KEY (id_modalidad) REFERENCES modalidades(id_modalidad)
    );
    """
]

# Ejecutar todas las queries
for query in queries:
    cur.execute(query)

conn.commit()
cur.close()
conn.close()

print("Tablas creadas exitosamente.")
```
