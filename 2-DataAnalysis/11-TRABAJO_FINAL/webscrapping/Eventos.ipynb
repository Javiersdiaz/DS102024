```python
# Tratado de datos

import pandas as pd
import numpy as np 

# construccion de datos

import random
from datetime import datetime
from faker import Faker

# Exportacion

from sqlalchemy import create_engine,inspect
```

# Utiles

```python
# Generamos los datos ficticios en espanhol

fake = Faker('es_ES')
```

#  Dataset de eventos

```python
# Creamos nuestros datos a utilizar en variables para el dataset de eventos 

id_evento = range(1,501,1)

nombre_evento = ['Pintura divertida','Juegos al sol',
                 'Exploradores','Fiesta de burbujas',
                 'Saltos y risas']

ubicacion =  fake.city()

fecheas = fake.date_between(start_date=datetime(2025, 1, 1), end_date=datetime(2025, 12, 31))

categoria = ['Talleres', 'Salud', 
             'Deportes', 'Cultura',
             'Cine','Viajes']

discapacidad = ['auditiva','física',
                'motora','intelectual',
                'ninguna']

integrantes = random.randint(3, 20)

edades_dirigido = ['0-4','5-8', '9-12', '13-16', '17-20', '21+']

modalidades = ['Interior', 'Aire libre']

precio = round(random.uniform(0, 100)),
```

```python
# creamos nuestro dataset con un bucle

eventos = []

for i in id_evento:

    nombre = random.choice(nombre_evento)

    if nombre == 'Pintura divertida':
        modalidad = 'Interior'
    elif nombre == 'Fiesta de burbujas':
        modalidad = 'Interior'
    else:
        modalidad = 'Aire libre'


    dicc = {'id_evento' : i,
            'nombre_evento':nombre,
            'ubicacion':fake.city(),
            'fechea':fake.date_between(start_date=datetime(2025, 1, 1), end_date=datetime(2025, 12, 31)),
            'categoria':random.choice(categoria),
            'discapacidad':random.choice(discapacidad),
            'integrantes':random.randint(3, 20),
            'edades':random.choice(edades_dirigido),
            'modalidades':modalidad,
            'precio':round(random.uniform(0, 100)),}
    
    eventos.append(dicc)

df_eventos = pd.DataFrame(eventos)
```

# Dataset de usuarios

```python
# Creamos nuestros datos a utilizar en variables para el dataset de usuarios

id_usuario = range(1,201,1)

discapacidad = ['auditiva','física',
                'motora','intelectual',
                'ninguna']

interes = ['Talleres', 'Salud', 
             'Deportes', 'Cultura',
             'Cine','Viajes']

edad_familiares = ['0-4','5-8',
                   '9-12','13-16',
                   '17-20', '21+']
```

```python
usuarios = []

for i in id_usuario:

    dicc = {'id_usuario': i,
            'nombre_usuario': fake.name(),
            'edad_usuario': random.choice(range(15,40)),
            'edad_familiares': random.choice(edad_familiares),
            'integrantes': random.choice(range(2,5)),
            'discapacidad':random.choice(discapacidad),
            'interes': random.choice(interes),
            'ubicacion':fake.city(),}
    usuarios.append(dicc)

df_usuarios = pd.DataFrame(usuarios)
```

# Dataset de registros

```python
id_historial_usuario = range(1,101,1)
```

```python
historial_usuario = []

for i in id_historial_usuario:
    dicc = {'id_historial_usuario': i,
            'id_usuario': random.choice(df_usuarios['id_usuario']),
            'id_evento': random.choice(df_eventos['id_evento']),
            'interaccion': random.choice([True, False])}
    
    historial_usuario.append(dicc)

df_historial_usuario = pd.DataFrame(historial_usuario)
```
