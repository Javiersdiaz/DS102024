```python
CREATE TABLE FAMILIAS (
    id_familia INT,
    email VARCHAR(255),
    nombre_contacto VARCHAR(255),
    apellido_contacto VARCHAR(255),
    telefono VARCHAR(255),
    direccion VARCHAR(255),
    codigo_postal VARCHAR(255),
    ciudad VARCHAR(255),
    provincia VARCHAR(255),
    fecha_registro TIMESTAMP
);

CREATE TABLE NIÑOS (
    id_nino INT,
    id_familia INT,
    nombre VARCHAR(255),
    fecha_nacimiento DATE,
    genero VARCHAR(255),
    intereses TEXT
);

CREATE TABLE INSCRIPCIONES (
    id_usuario INT,
    id_actividad INT,
    fecha_inscripcion TIMESTAMP
);

CREATE TABLE INTERESES (
    id_interes INT,
    nombre VARCHAR(255),
    descripcion TEXT
);

CREATE TABLE ACTIVIDADES (
    id_actividad INT,
    nombre VARCHAR(255),
    descripcion_corta TEXT,
    descripcion_larga TEXT,
    tipo VARCHAR(255),
    id_categoria INT,
    id_necesidad INT,
    edad_minima INT,
    edad_maxima INT,
    id_ubicacion INT,
    id_temporada INT,
    id_coste INT,
    id_entorno INT,
    rango_integrantes_min INT,
    rango_integrantes_max INT,
    link VARCHAR(255),
    imagen_principal VARCHAR(255),
    fecha_inicio DATE,
    fecha_fin DATE,
    hora_inicio TIME,
    hora_fin TIME,
    organizador VARCHAR(255),
    telefono_contacto VARCHAR(255),
    email_contacto VARCHAR(255)
);

CREATE TABLE CATEGORIAS (
    id_categoria INT,
    nombre VARCHAR(255),
    descripcion TEXT
);

CREATE TABLE NECESIDADES (
    id_necesidad INT,
    nombre VARCHAR(255),
    descripcion TEXT
);

CREATE TABLE EDADES (
    id_edad INT,
    rango VARCHAR(255),
    edad_minima INT,
    edad_maxima INT
);

CREATE TABLE UBICACIONES (
    id_ubicacion INT,
    nombre VARCHAR(255),
    latitud DECIMAL(10,6),
    longitud DECIMAL(10,6)
);

CREATE TABLE TEMPORADAS (
    id_temporada INT,
    nombre VARCHAR(255)
);

CREATE TABLE COSTES (
    id_coste INT,
    nombre VARCHAR(255),
    valor_minimo DECIMAL(10,6),
    valor_maximo DECIMAL(10,6)
);

CREATE TABLE ENTORNOS (
    id_entorno INT,
    nombre VARCHAR(255),
    descripcion TEXT
);

CREATE TABLE FUENTES_DATOS (
    id_fuente INT,
    nombre VARCHAR(255),
    url_base VARCHAR(255),
    tipo_fuente VARCHAR(255),
    fecha_ultimo_scrapping TIMESTAMP,
    configuracion_scrapping TEXT
);

CREATE TABLE PLANES_CRUDOS (
    id_plan_crudo INT,
    id_fuente INT,
    url_plan VARCHAR(255),
    titulo TEXT,
    descripcion TEXT,
    fecha_extraccion TIMESTAMP,
    datos_extraidos JSON
);
```
